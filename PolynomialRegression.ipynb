```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
```

```python
speed = np.array([20,40,60,80,100,120]).reshape(-1,1)
brake = np.array([3,10,22,40,62,90])
```

```python
model = LinearRegression()
```

```python
model.fit(speed,brake)
```

```python
pred = model.predict(speed)
```

```python
plt.scatter(speed,brake,color='blue',label='Actual Values')
plt.plot(speed,pred,color='red',label='Predicted Value')
plt.xlabel('Speed')
plt.ylabel('Brakes')
plt.legend()
plt.title('Non Curved')
plt.show()
```

```python
poly = PolynomialFeatures(degree=2)
```

```python
poly_speed = poly.fit_transform(speed,brake)
```

```python
poly_model = LinearRegression()
```

```python
poly_model.fit(poly_speed,brake)
```

```python
poly_pred = poly_model.predict(poly_speed)
```

```python
plt.scatter(speed,brake,color='blue',label='Actual')
plt.plot(speed,poly_pred,color='red',label='Predicted')
plt.xlabel('Speed')
plt.ylabel('Brake')
plt.legend()
plt.title('Curved')
```

```python
highpoly = PolynomialFeatures(degree=8)
```

```python
highspeed = highpoly.fit_transform(speed,brake)
```

```python
highmodel = LinearRegression()
```

```python
highmodel.fit(highspeed,brake)
```

```python
highpred = highmodel.predict(highspeed)
```

```python
plt.scatter(speed,brake,color='blue',label='Actual')
plt.plot(speed,highpred,color='red',label='Predicted')
plt.xlabel('Speed')
plt.ylabel('Brake')
plt.legend()
plt.title('Too much Curved')
plt.show()
```
