```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
units = np.array([28,101,106,117,127,134])
bill = np.array([560,12300,14000,16000,21000,30100])
```

```python
model = LinearRegression()
```

```python
model.fit(units.reshape(-1,1),bill)
```

```python
your_unit = float(input("Enter your units to get the bill : "))
prediction = model.predict([[your_unit]])
print(prediction)
```

```python
plt.plot(units,bill)
plt.scatter(units,model.predict(units.reshape(-1,1)),color='red')
```
