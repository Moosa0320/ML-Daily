```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Har row: [age, km, transmission, petrol, diesel, hybrid]
cars = np.array([
    [2, 20000, 1, 1, 0, 0],
    [5, 60000, 0, 0, 1, 0],
    [3, 35000, 1, 0, 0, 1],
    [8, 90000, 0, 1, 0, 0],
    [1, 10000, 1, 0, 0, 1]
])

price = np.array([18, 10, 15, 6, 20])
```

```python
model = LinearRegression()
```

```python
model.fit(cars,price)
```

```python
new_car = [[4,4000,0,1,0,0]]
pre = model.predict(new_car) 
print(pre)          
```
