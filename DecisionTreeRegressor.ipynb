```python
import numpy as np
from sklearn.tree import DecisionTreeRegressor, plot_tree
import matplotlib.pyplot as plt
```

```python
X = np.array([
    [500, 1, 5],
    [750, 2, 3],
    [1000, 2, 8],
    [1200, 3, 2],
    [1500, 3, 10],
    [1800, 4, 1],
    [2000, 4, 6]
])

rent = np.array([15000, 22000, 25000, 35000, 38000, 50000, 52000])
```

```python
model = DecisionTreeRegressor(max_depth=3,random_state=42)
model.fit(X,rent)
```

```python
new_house = [[1300, 3, 4]]
predicted_rent = model.predict(new_house)
print("Predicted Rent:", predicted_rent[0])
```

```python
plt.Figure(figsize=(25,17))
plot_tree(model,feature_names=['Size','Bedroom','Age'],filled=True)
plt.show()
```

```python
importance = model.feature_importances_
features = ['Size','Bedroom','Age']

for feat, imp in zip(features, importance):
    print(f"{feat} : {imp:.3f}")
```

```python
from sklearn.metrics import mean_squared_error
```

```python
pred_all = model.predict(X)
error = mean_squared_error(rent,pred_all)
print(error)
```

```python
plt.scatter(rent, pred_all, color='blue')
plt.plot([min(rent), max(rent)], [min(rent), max(rent)], color='red')
plt.xlabel("Actual Rent")
plt.ylabel("Predicted Rent")
plt.title("Regression Tree: Actual vs Predicted")
plt.show()
```
